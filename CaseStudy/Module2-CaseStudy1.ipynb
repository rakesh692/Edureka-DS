## A website requires a user to input a username and password to register. Write a program to check the validity of the password given by the user. Following are the criteria for checking password:
</br> 1. At least 1 letter between [a-z]
</br> 2. At least 1 number between [0-9]
</br> 3. At least 1 letter between [A-Z]
</br> 4. At least 1 character from [$#@]
</br> 5. Minimum length of transaction password: 6
</br> 6. Maximum length of transaction password: 12

```python
import re

def validate_password(password):
    # Check length
    if len(password) < 6 or len(password) > 12:
        return "Invalid: Password length must be between 6 and 12 characters."

    # Check for lowercase letter
    if not re.search("[a-z]", password):
        return "Invalid: Must contain at least one lowercase letter."

    # Check for uppercase letter
    if not re.search("[A-Z]", password):
        return "Invalid: Must contain at least one uppercase letter."

    # Check for digit
    if not re.search("[0-9]", password):
        return "Invalid: Must contain at least one digit."

    # Check for special character
    if not re.search("[$#@]", password):
        return "Invalid: Must contain at least one special character ($#@)."

    return "Valid Password!"

# Example usage:
user_password = input("Enter your password: ")
print(validate_password(user_password))
```

## Please write a program thatcountsand printsthe numbers of each character in a string input by the console

```python

# Program to count occurrences of each character in a string
user_input = input("Enter a string: ")

# Dictionary to store character counts
char_count = {}

for char in user_input:
    char_count[char] = char_count.get(char, 0) + 1

# Print the counts
print("Character counts:")
for char, count in char_count.items():
    print(f"'{char}' : {count}")
```

## With   two   given   lists   [1,3,6,78,35,55]   and   [12,24,35,24,88,120,155],   write   a program to make a list whose elements are intersection of the above given lists.

```python

# Original list
numbers = [12, 24, 35, 70, 88, 120, 155]

# Remove numbers divisible by both 5 and 7 using list comprehension
filtered_list = [x for x in numbers if not (x % 5 == 0 and x % 7 == 0)]

print("List after removing numbers divisible by both 5 and 7:", filtered_list)
```

## Write  a  program  to  compute  1/2+2/3+3/4+...+n/n+1  with  a  given  n  input  by console (n>0).

```python

# Program to compute 1/2 + 2/3 + 3/4 + ... + n/(n+1)

# Take input from user
n = int(input("Enter a positive integer n: "))

if n > 0:
    total = sum(i / (i + 1) for i in range(1, n + 1))
    print(f"The sum of the series for n = {n} is: {total}")
else:
    print("Please enter a number greater than 0.")
```
