# Merge

```python
df5 = pd.DataFrame({"ID":[1,2,3,4],"Name":["Alice","Bob","Charlie","David"],"Age":[25,30,35,40]})
df5
```

```python
df6 = pd.DataFrame({"ID":[3,4,5,6],"Name":["Charlie","David","Eve","Frank"],"Age":[35,40,50,70]})
df6
```

```python
innerj = pd.merge(df5,df6,on="ID",how="inner")
innerj
```

```python
innerj_clean = innerj.drop(["Name_y","Age_y"],axis=1)
innerj_clean
```

```python
innerj_clean.columns = ["ID","Name","Age"]
innerj_clean
```

```python
leftj = pd.merge(df5,df6,on="ID",how="left")
leftj
```

```python
rightj = pd.merge(df5,df6,on="ID",how="right")
rightj
```

```python
outerj = pd.merge(df5,df6,on="ID",how="outer")
outerj
```

```python
#data : df
# values : the column names for aggregation
# index : column name in row
# columns : column name in column
# aggfunc : what function to apply on the value column
```

```python
data = {"Region":["East","West","East","North","East","South"],
       "Product":["A","B","A","B","C","A"],
       "Sales":[100,200,150,400,500,100]}

df = pd.DataFrame(data)
df
```

```python
sales_summary = pd.pivot_table(df,values="Sales",index="Region",columns="Product",aggfunc="sum")
sales_summary
```

```python
sales_summary.fillna(0)
```

```python
data2 = {"Region":["East","West","East","North","East","South"],
       "Product":["A","B","A","B","C","A"],
       "Sales":[100,200,150,400,500,100],
        "Profit":[10,20,15,40,50,10]}

df2 = pd.DataFrame(data2)
df2
```

```python
# multiple aggregations:

summary = pd.pivot_table(df2,values=["Sales","Profit"],index="Region",aggfunc={"Sales":["sum","mean"],"Profit":"max"})
summary
```

```python
#multi-level indexing

summary = pd.pivot_table(df2,values=["Sales","Profit"],index=["Region","Product"],aggfunc="sum",margins=True,margins_name="Total")
summary
```

# Iterating over a dataframe

```python
d = pd.DataFrame(np.random.rand(5,4),columns=["C1","C2","C3","C4"])
d
```

```python
# Method 1: Via .items()
for k,v in d.items():
    print("Key:",k)
    print("Value is:")
    print(v)
```

```python
# Method 2: Via .iterrows()
for k,v in d.iterrows():
    print("Key:",k)
    print("Value is:")
    print(v)
```

```python
# Method 3: Via .itertuples()
for t in d.itertuples():
    print(t)
```
