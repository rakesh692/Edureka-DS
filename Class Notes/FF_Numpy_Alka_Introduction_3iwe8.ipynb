```python
import numpy as np
data = np.array([10,14,15,18,21,23,30,35])
data
```

```python
q1 = np.percentile(data,25)
print("Q1:",q1)
q2 = np.percentile(data,50)
print("Q2:",q2)
q3 = np.percentile(data,75)
print("Q3:",q3)
```

```python
# Pandas with 'midpoint' method (closer to manual split-halves definition)

import pandas as pd

data = pd.Series([10,14,15,18,21,23,30,35])

q1 = data.quantile(0.25, interpolation="midpoint")
q3 = data.quantile(0.75, interpolation="midpoint")

print("Q1:", q1, "Q3:", q3)
```
