```python
# strings are immutable
s = "abc"
print("Original string:", s, "ID:", id(s))
new_s = s.replace("p", "P")
print("After replace():", new_s, "ID:", id(new_s))
```

```python
# Numbers are immutable
x = 10
print("Original number:", x)
print("ID before change:", id(x))

x = x + 5  # Perform addition
print("Modified number:", x)
print("ID after change:", id(x))
```
